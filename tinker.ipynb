```python
import pandas as pd

df = pd.read_excel('chores_centralized.xlsx', sheet_name='completions')
df.head()
```

```python
done_count_per_person = df[df['status'] == 'done'].groupby('person_id').size()
total_count_per_person = df.groupby('person_id').size()
house_done_p = done_count_per_person.sum() / total_count_per_person.sum()

print("Done count per person:\n\n", done_count_per_person, "\n")
print("Total count per person:\n\n", total_count_per_person, "\n")
print(f"House done percentage: {house_done_p:.2%}\n")
```

```python
k = 10
bayes_per_person = (done_count_per_person + k * house_done_p) / (total_count_per_person + k) 
bayes_per_person = bayes_per_person.sort_values(ascending=False)

for person_id, p in bayes_per_person.items():
    print(f"{person_id} Bayes perc: {p:.2%}")
```
